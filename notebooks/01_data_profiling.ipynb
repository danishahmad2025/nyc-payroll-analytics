## Initial Observation

The dataset contains:
- numerical columns
- categorical columns
- salary-related columns

```python
#creating optimized data type dictionary for the columns

dtype_dict = {
    "Agency Name": "category",
    "Work Location Borough": "category",
    "Title Description": "category",
    "Pay Basis": "category"
}
```

```python
import pandas as pd
chunk_size = 100000
chunks = pd.read_csv(
    file_path,
    dtype=dtype_dict,
    chunksize=chunk_size,
    low_memory=False
)
```

```python
# initializing profiling variables
total_rows = 0
null_counts = None
duplicate_count = 0
salary_sum = 0
salary_count = 0

for chunk in chunks:
    
    # Count rows
    total_rows += len(chunk)
    
    # Null values
    if null_counts is None:
        null_counts = chunk.isnull().sum()
    else:
        null_counts += chunk.isnull().sum()
    
    # Duplicate rows
    duplicate_count += chunk.duplicated().sum()
    
    
    chunk["Base Salary"] = (
        chunk["Base Salary"]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("$", "", regex=False)
        .str.strip()
    )
    #converting Base Salary to numeric.
    chunk["Base Salary"] = pd.to_numeric(chunk["Base Salary"], errors="coerce")
    # Salary statistics
    salary_sum += chunk["Base Salary"].sum()
    
    salary_count += chunk["Base Salary"].count()
    
if salary_count > 0:
    average_salary = salary_sum / salary_count
else:
    average_salary = 0
print("Total Rows:", total_rows)

print("\nNull Values:")
print(null_counts)

print("\nDuplicate Rows:")
print(duplicate_count)

print("\nAverage Salary:")
print(round(average_salary, 2))
```
